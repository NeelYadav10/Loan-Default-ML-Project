### 2.2 Missing Value Strategy & Justification
Even though this dataset is clean, robust pipelines require explicit missing value handling:
- **Numerical Features:** Imputed using **Median** via `SimpleImputer(strategy='median')`. Median imputation is robust against skewed numerical distributions and extreme outliers.
- **Categorical Features:** Imputed using **Mode** via `SimpleImputer(strategy='most_frequent')`. Mode imputation preserves valid discrete category levels without introducing artificial non-integer values.

```python
# Step 2.2: Apply SimpleImputer for numerical and categorical features
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.drop('Default', errors='ignore').tolist()
if 'LoanID' in num_cols:
    num_cols.remove('LoanID')
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
if 'LoanID' in cat_cols:
    cat_cols.remove('LoanID')

if len(num_cols) > 0 and df[num_cols].isnull().sum().sum() > 0:
    num_imputer = SimpleImputer(strategy='median')
    df[num_cols] = num_imputer.fit_transform(df[num_cols])

if len(cat_cols) > 0 and df[cat_cols].isnull().sum().sum() > 0:
    cat_imputer = SimpleImputer(strategy='most_frequent')
    df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])

print("Imputation pipeline executed. Total missing values remaining:", df.isnull().sum().sum())
```

### 2.3 Outlier Detection & Handling (IQR Method)
The **Interquartile Range (IQR)** method identifies extreme numerical anomalies beyond statistical thresholds:
$$IQR = Q_3 - Q_1$$
$$\text{Lower Bound} = Q_1 - 1.5 \times IQR, \quad \text{Upper Bound} = Q_3 + 1.5 \times IQR$$

Rather than dropping valuable borrower records, extreme outliers beyond these bounds are **capped** to upper and lower limits to retain total sample size while neutralizing leverage effects.

```python
# Visualize boxplots of key numerical features BEFORE outlier capping
key_num_cols = ['Age', 'Income', 'LoanAmount', 'CreditScore', 'MonthsEmployed', 'InterestRate', 'DTIRatio']
existing_cols = [c for c in key_num_cols if c in df.columns]

plt.figure(figsize=(15, 8))
for i, col in enumerate(existing_cols, 1):
    plt.subplot(2, 4, i)
    sns.boxplot(y=df[col], color='skyblue')
    plt.title(f"Before IQR: {col}", fontsize=11, fontweight='bold')
plt.suptitle("Numerical Feature Distributions (BEFORE Outlier Capping)", y=1.02, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Apply IQR capping on numerical features
df_clean = df.copy()
capping_summary = {}

for col in existing_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers_below = (df_clean[col] < lower_bound).sum()
    outliers_above = (df_clean[col] > upper_bound).sum()
    capping_summary[col] = outliers_below + outliers_above
    
    df_clean[col] = np.clip(df_clean[col], lower_bound, upper_bound)

df = df_clean
print("Capped Outliers Count per Feature:", capping_summary)

plt.figure(figsize=(15, 8))
for i, col in enumerate(existing_cols, 1):
    plt.subplot(2, 4, i)
    sns.boxplot(y=df[col], color='lightgreen')
    plt.title(f"After IQR: {col}", fontsize=11, fontweight='bold')
plt.suptitle("Numerical Feature Distributions (AFTER Outlier Capping)", y=1.02, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 2.4 Categorical Encoding Strategy
- **LabelEncoder vs. OneHotEncoder:**

  - **OneHotEncoder:** Creates dummy indicator columns for nominal multi-class variables to avoid introducing false numerical ranking.
  - Here, `LabelEncoder` is applied across binary and categorical features to yield a compact matrix representation for modeling.

```python
# Step 2.4: Encode categorical variables using LabelEncoder
cat_columns = ['Education', 'EmploymentType', 'MaritalStatus', 'HasMortgage', 'HasDependents', 'LoanPurpose', 'HasCoSigner']
encoders = {}

for col in cat_columns:
    if col in df.columns:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col].astype(str))
        encoders[col] = le

print("Categorical Encoding complete. Sample encoded values:")
df[cat_columns].head()
```

### 3.1 Justification: Classification vs. Regression
Loan default prediction is inherently a **classification task** ($y \in \{0, 1\}$) rather than a regression problem for the following reasons:
1. **Discrete Categorical Target:** The target variable `Default` is qualitative and binary (0 = non-default, 1 = default).
2. **Bounded Output Range:** Linear regression models output continuous unbounded predictions ($y \in \mathbb{R}$) which can generate uninterpretable probabilities $<0$ or $>1$.
3. **Loss Function Optimization:** Classification models optimize log-loss (binary cross-entropy) or tree impurity (Gini/Entropy), matching the probability density of decision boundaries.

```python
# Step 3.1: Train Scikit-Learn Logistic Regression Model
t0 = time.time()
sk_model = LogisticRegression(max_iter=1000, random_state=42)
sk_model.fit(X_train_scaled, y_train)
sk_train_time = time.time() - t0

y_pred_sk = sk_model.predict(X_test_scaled)

sk_acc = accuracy_score(y_test, y_pred_sk)
sk_prec = precision_score(y_test, y_pred_sk, zero_division=0)
sk_rec = recall_score(y_test, y_pred_sk, zero_division=0)
sk_f1 = f1_score(y_test, y_pred_sk, zero_division=0)

print("=== Scikit-Learn Logistic Regression Results ===")
print(f"Training Time: {sk_train_time:.4f} seconds")
print(f"Accuracy:  {sk_acc:.4f}")
print(f"Precision: {sk_prec:.4f}")
print(f"Recall:    {sk_rec:.4f}")
print(f"F1-Score:  {sk_f1:.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_sk, zero_division=0))

cm_sk = confusion_matrix(y_test, y_pred_sk)
disp = ConfusionMatrixDisplay(confusion_matrix=cm_sk, display_labels=['No Default (0)', 'Default (1)'])
disp.plot(cmap='Blues')
plt.title("Scikit-Learn Logistic Regression - Confusion Matrix", fontweight='bold')
plt.grid(False)
plt.show()
```

### 3.2 Custom Implementation from Scratch (NumPy Logistic Regression)

The custom model implements **Logistic Regression with Gradient Descent** using pure NumPy:
- **Hypothesis / Sigmoid Activation:**
  $$\hat{y} = \sigma(z) = \frac{1}{1 + e^{-(X w + b)}}$$
- **Binary Cross-Entropy Cost Function:**
  $$J(w, b) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log(\hat{y}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{y}^{(i)}) \right]$$
- **Gradient Descent Updates:**
  $$w \leftarrow w - \alpha \frac{1}{m} X^T (\hat{y} - y), \quad b \leftarrow b - \alpha \frac{1}{m} \sum (\hat{y} - y)$$

```python
# Step 3.2: Complete Custom Logistic Regression Class in Pure NumPy
class LogisticRegressionFromScratch:
    def __init__(self, learning_rate=0.1, num_iterations=1000):
        self.learning_rate = learning_rate
        self.num_iterations = num_iterations
        self.weights = None
        self.bias = None
        self.cost_history = []

    def _sigmoid(self, z):
        z_clipped = np.clip(z, -250, 250)
        return 1.0 / (1.0 + np.exp(-z_clipped))

    def _compute_cost(self, y_true, y_pred):
        m = len(y_true)
        epsilon = 1e-15
        y_pred_clipped = np.clip(y_pred, epsilon, 1.0 - epsilon)
        cost = -(1 / m) * np.sum(y_true * np.log(y_pred_clipped) + (1 - y_true) * np.log(1.0 - y_pred_clipped))
        return cost

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0
        y_arr = y.to_numpy() if hasattr(y, 'to_numpy') else np.array(y)

        for i in range(self.num_iterations):
            linear_model = np.dot(X, self.weights) + self.bias
            y_pred = self._sigmoid(linear_model)

            dw = (1 / n_samples) * np.dot(X.T, (y_pred - y_arr))
            db = (1 / n_samples) * np.sum(y_pred - y_arr)

            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

            if i % 100 == 0:
                c = self._compute_cost(y_arr, y_pred)
                self.cost_history.append(c)

    def predict_proba(self, X):
        linear_model = np.dot(X, self.weights) + self.bias
        return self._sigmoid(linear_model)

    def predict(self, X, threshold=0.5):
        y_predicted = self.predict_proba(X)
        return (y_predicted >= threshold).astype(int)

# Train From-Scratch Model
t0 = time.time()
scratch_model = LogisticRegressionFromScratch(learning_rate=0.1, num_iterations=1000)
scratch_model.fit(X_train_scaled, y_train)
scratch_train_time = time.time() - t0

y_pred_scratch = scratch_model.predict(X_test_scaled)

scratch_acc = accuracy_score(y_test, y_pred_scratch)
scratch_prec = precision_score(y_test, y_pred_scratch, zero_division=0)
scratch_rec = recall_score(y_test, y_pred_scratch, zero_division=0)
scratch_f1 = f1_score(y_test, y_pred_scratch, zero_division=0)

print("=== From-Scratch Logistic Regression Results ===")
print(f"Training Time: {scratch_train_time:.4f} seconds")
print(f"Accuracy:  {scratch_acc:.4f}")
print(f"Precision: {scratch_prec:.4f}")
print(f"Recall:    {scratch_rec:.4f}")
print(f"F1-Score:  {scratch_f1:.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_scratch, zero_division=0))

cm_scratch = confusion_matrix(y_test, y_pred_scratch)
disp_scratch = ConfusionMatrixDisplay(confusion_matrix=cm_scratch, display_labels=['No Default (0)', 'Default (1)'])
disp_scratch.plot(cmap='Greens')
plt.title("From-Scratch Logistic Regression - Confusion Matrix", fontweight='bold')
plt.grid(False)
plt.show()
```

### 3.3 Performance Comparison & Discussion

1. **Accuracy & Metric Parity:**
   - Both models achieve virtually identical accuracy (**~88.51%** vs **~88.50%**) on the test dataset.
   - Their precision, recall, and F1-scores align closely, proving that the mathematical formulation of our custom NumPy gradient descent implementation matches Scikit-Learn's underlying optimization logic.

2. **Execution Speed & Computational Efficiency:**
   - **Scikit-Learn** trained in **~0.16 seconds**.
   - **From-Scratch NumPy** trained in **~7.5 seconds** (nearly **45x slower**).
   - *Reason:* Scikit-Learn wraps highly optimized C/Fortran routines (`LIBLINEAR` / `L-BFGS`) and utilizes quasi-Newton second-order Hessian approximations, whereas pure Python/NumPy gradient descent performs un-parallelized first-order updates step-by-step.

3. **Why Machine Learning Libraries are Used in Practice:**
   - **Numerical Stability:** Production libraries prevent gradient explosion and floating-point underflow/overflow using techniques like log-sum-exp stabilization and dynamic learning rate line searches.
   - **Advanced Optimization Solvers:** Solvers like L-BFGS, SAG, and SAGA converge in fewer iterations than standard first-order gradient descent.
   - **Built-in Regularization & Hyperparameter Tuning:** Automatic $L_1$ and $L_2$ penalty handling prevents overfitting without custom gradient algebra.
   - **Scalability & Production Infrastructure:** Integration with multi-threading, C++ backends, pipelines, and deployment frameworks makes libraries indispensable for industry applications.

## Week 4 — Task 5: Model Evaluation, Comparison & Tuning

In this section, we expand beyond our baseline linear models by implementing and evaluating advanced tree-based ensemble architectures: **Random Forest (Bagging)**, **AdaBoost (Adaptive Boosting)**, and **Gradient Boosting**. We systematically conduct four diagnostic evaluations: (1) **Comprehensive Multi-Metric Evaluation** (Accuracy, Precision, Recall, F1-Score) on the held-out test set; (2) **Overfitting vs. Underfitting Diagnostics** by analyzing the generalization gap between train and test performance; (3) **5-Fold Stratified Cross-Validation** to measure generalization stability and flag high variance; and (4) an integrated **Model Comparison Table** benchmarking all five classification models to select the champion architecture. Finally, we apply **GridSearchCV** hyperparameter optimization on the champion model to maximize F1-score and verify test-set performance improvements.


### 5.1 Training Advanced Classification Models (Random Forest, AdaBoost, Gradient Boosting)

Linear models often struggle with complex, non-linear feature interactions in tabular loan data. We train three distinct ensemble algorithms:
1. **Random Forest Classifier (Bagging):** Builds an ensemble of de-correlated decision trees in parallel using bootstrap aggregation and random feature sub-sampling, reducing variance without inflating bias.
2. **AdaBoost Classifier (Adaptive Boosting):** Sequentially trains decision stumps, iteratively upweighting previously misclassified borrower instances to focus learning on difficult boundary cases.
3. **Gradient Boosting Classifier (Gradient Boosting):** Sequentially fits regression trees to the negative gradients (pseudo-residuals) of the binary cross-entropy loss function via gradient descent in function space.


```python
# Step 5.1: Train Advanced Ensemble Models and Compute Test-Set Metrics
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier
from sklearn.model_selection import StratifiedKFold, cross_val_score, GridSearchCV

# Initialize ensemble models
rf_model = RandomForestClassifier(n_estimators=100, max_depth=12, random_state=42, n_jobs=-1)
ada_model = AdaBoostClassifier(n_estimators=100, random_state=42)
gb_model = GradientBoostingClassifier(n_estimators=100, max_depth=4, learning_rate=0.1, random_state=42)

ensemble_models = {
    'Random Forest (Bagging)': rf_model,
    'AdaBoost': ada_model,
    'Gradient Boosting': gb_model
}

ensemble_metrics = {}

for name, model in ensemble_models.items():
    t0 = time.time()
    model.fit(X_train_scaled, y_train)
    fit_time = time.time() - t0
    
    y_pred = model.predict(X_test_scaled)
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    
    ensemble_metrics[name] = {
        'model': model,
        'y_pred': y_pred,
        'time': fit_time,
        'Accuracy': acc,
        'Precision': prec,
        'Recall': rec,
        'F1-score': f1
    }
    
    print(f"=== {name} Results ===")
    print(f"Training Time: {fit_time:.4f} seconds")
    print(f"Accuracy:  {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall:    {rec:.4f}")
    print(f"F1-Score:  {f1:.4f}")
    print(f"\nClassification Report:\n", classification_report(y_test, y_pred, zero_division=0))
    print("-" * 55)

# Visualize Confusion Matrices for the three ensemble architectures
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
cm_palettes = ['Blues', 'Purples', 'Oranges']

for i, (name, data) in enumerate(ensemble_metrics.items()):
    cm = confusion_matrix(y_test, data['y_pred'])
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['No Default (0)', 'Default (1)'])
    disp.plot(cmap=cm_palettes[i], ax=axes[i], colorbar=False)
    axes[i].set_title(f"{name}\nConfusion Matrix", fontweight='bold', fontsize=12)
    axes[i].grid(False)

plt.tight_layout()
plt.show()
```

### 5.2 Overfitting vs. Underfitting Diagnostics

To diagnose the bias-variance tradeoff across all models, we evaluate training accuracy against test accuracy using strict generalization criteria:
- **Overfitting (High Variance):** Train Score $\gg$ Test Score (Generalization gap $\Delta > 5\%$), indicating the model learned sample noise rather than structural patterns.
- **Underfitting (High Bias):** Both Train and Test scores are poor ($< 70\%$), indicating insufficient model expressive power.
- **Good Fit (Balanced Generalization) ✅:** Train and Test scores are closely aligned (Generalization gap $\Delta \le 5\%$) with high baseline accuracy ($\\ge 85\%$).


```python
# Step 5.2: Overfitting / Underfitting Check for All 5 Models
all_models_dict = {
    'Scikit-Learn Logistic Regression': sk_model,
    'From-Scratch Logistic Regression': scratch_model,
    'Random Forest': rf_model,
    'AdaBoost': ada_model,
    'Gradient Boosting': gb_model
}

fit_check_results = []

for name, model in all_models_dict.items():
    if hasattr(model, 'score'):
        train_acc = model.score(X_train_scaled, y_train)
        test_acc = model.score(X_test_scaled, y_test)
    else:
        # For custom from-scratch model
        train_acc = accuracy_score(y_train, model.predict(X_train_scaled))
        test_acc = accuracy_score(y_test, model.predict(X_test_scaled))
    
    diff = train_acc - test_acc
    
    # Diagnostic classification rule
    if diff > 0.05:
        diagnosis = 'Overfitting (Train >> Test) ⚠️'
    elif train_acc < 0.70 and test_acc < 0.70:
        diagnosis = 'Underfitting (Both Low) ⚠️'
    else:
        diagnosis = 'Good fit (Train ≈ Test) ✅'
        
    fit_check_results.append({
        'Model': name,
        'Train Accuracy': train_acc,
        'Test Accuracy': test_acc,
        'Generalization Gap (Train - Test)': diff,
        'Diagnosis': diagnosis
    })

fit_df = pd.DataFrame(fit_check_results)
print("=== Overfitting / Underfitting Diagnostic Summary ===")
try:
    display(fit_df)
except NameError:
    print(fit_df.to_string(index=False))

# Visualize Train vs. Test Scores
plt.figure(figsize=(12, 5))
x_indices = np.arange(len(fit_df))
width = 0.35

plt.bar(x_indices - width/2, fit_df['Train Accuracy'], width, label='Train Accuracy', color='steelblue', alpha=0.9)
plt.bar(x_indices + width/2, fit_df['Test Accuracy'], width, label='Test Accuracy', color='coral', alpha=0.9)

plt.title("Overfitting / Underfitting Diagnostic: Train vs. Test Accuracy", fontsize=13, fontweight='bold')
plt.xlabel("Model Architecture", fontweight='bold')
plt.ylabel("Accuracy Score", fontweight='bold')
plt.xticks(x_indices, fit_df['Model'], rotation=15, ha='right')
plt.ylim(0.80, 1.0)
plt.legend(frameon=True)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### 5.3 5-Fold Stratified Cross-Validation & Stability Assessment

Single train-test splits can yield optimistic or pessimistic performance estimates due to sampling variability. We apply **5-Fold Stratified Cross-Validation** ($k=5$) to:
1. Maintain the target class distribution (~88:12) consistently across all five validation folds.
2. Calculate the **Mean F1-Score** (expected generalization capability).
3. Compute the **Score Spread / Standard Deviation ($\sigma$)** across folds: a small $\sigma$ ($< 0.015$) confirms model stability, while a high spread ($\\sigma \ge 0.015$) indicates sensitivity to data partitioning.


```python
# Step 5.3: 5-Fold Stratified Cross-Validation on Training Data
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_results = {}

print("=== 5-Fold Stratified Cross-Validation Results (Metric: F1-Score) ===")
for name, model in all_models_dict.items():
    if name == 'From-Scratch Logistic Regression':
        # Custom cross-validation loop for custom NumPy estimator
        scratch_cv_scores = []
        for train_idx, val_idx in cv.split(X_train_scaled, y_train):
            X_fold_tr, X_fold_val = X_train_scaled[train_idx], X_train_scaled[val_idx]
            y_fold_tr, y_fold_val = y_train.iloc[train_idx], y_train.iloc[val_idx]
            temp_scratch = LogisticRegressionFromScratch(learning_rate=0.1, num_iterations=1000)
            temp_scratch.fit(X_fold_tr, y_fold_tr)
            y_fold_pred = temp_scratch.predict(X_fold_val)
            scratch_cv_scores.append(f1_score(y_fold_val, y_fold_pred, zero_division=0))
        scores = np.array(scratch_cv_scores)
    else:
        scores = cross_val_score(model, X_train_scaled, y_train, cv=cv, scoring='f1', n_jobs=-1)
    
    cv_mean = scores.mean()
    cv_std = scores.std()
    cv_results[name] = {'scores': scores, 'mean': cv_mean, 'std': cv_std}
    
    # Flag high spread as instability warning
    stability_flag = 'Stable (Low Spread) ✅' if cv_std < 0.015 else 'Warning: High Spread ⚠️'
    
    print(f"{name}:")
    print(f"  Fold Scores: {[round(s, 4) for s in scores]}")
    print(f"  CV Mean F1:  {cv_mean:.4f} | CV Std (Spread): {cv_std:.4f} -> {stability_flag}")
    print("-" * 65)

# Visualize Cross-Validation F1-score distributions across folds
cv_plot_df = pd.DataFrame({k: v['scores'] for k, v in cv_results.items()})
plt.figure(figsize=(12, 5))
sns.boxplot(data=cv_plot_df, palette='Set2')
plt.title("5-Fold Cross-Validation F1-Score Distribution per Model", fontsize=13, fontweight='bold')
plt.xlabel("Model Architecture", fontweight='bold')
plt.ylabel("F1-Score", fontweight='bold')
plt.xticks(rotation=15, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### 5.5 Hyperparameter Tuning via GridSearchCV on Champion Model

We optimize the champion **Gradient Boosting** model using `GridSearchCV` with **5-Fold Stratified Cross-Validation** targeting `scoring='f1'`. Key hyperparameters tuned include:
- `n_estimators`: Total boosting stages (trees) constructed.
- `learning_rate`: Shrinkage parameter scaling the contribution of each individual tree.
- `max_depth`: Maximum depth of individual regression trees, governing interaction complexity.

Following grid optimization, the best estimator is re-evaluated on the unseen test set (`X_test_scaled`, `y_test`) to confirm performance improvements over the untuned baseline.


```python
# Step 5.5: Hyperparameter Tuning for GradientBoostingClassifier using GridSearchCV
param_grid = {
    'n_estimators': [100, 150],
    'learning_rate': [0.1, 0.15],
    'max_depth': [3, 4]
}

gb_base = GradientBoostingClassifier(random_state=42)
cv_strat = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

print("=== Initiating GridSearchCV for Gradient Boosting ===")
grid_search = GridSearchCV(
    estimator=gb_base,
    param_grid=param_grid,
    cv=cv_strat,
    scoring='f1',
    n_jobs=-1,
    verbose=1
)

t0 = time.time()
grid_search.fit(X_train_scaled, y_train)
tuning_duration = time.time() - t0

print(f"GridSearchCV completed in {tuning_duration:.2f} seconds.")
print(f"Best Hyperparameter Combination: {grid_search.best_params_}")
print(f"Best 5-Fold Cross-Validation F1-Score: {grid_search.best_score_:.4f}")

# Re-evaluate the tuned model on the held-out test set
best_gb_model = grid_search.best_estimator_
y_pred_tuned = best_gb_model.predict(X_test_scaled)

tuned_acc = accuracy_score(y_test, y_pred_tuned)
tuned_prec = precision_score(y_test, y_pred_tuned, zero_division=0)
tuned_rec = recall_score(y_test, y_pred_tuned, zero_division=0)
tuned_f1 = f1_score(y_test, y_pred_tuned, zero_division=0)

# Untuned baseline metrics for Gradient Boosting
untuned_acc = ensemble_metrics['Gradient Boosting']['Accuracy']
untuned_prec = ensemble_metrics['Gradient Boosting']['Precision']
untuned_rec = ensemble_metrics['Gradient Boosting']['Recall']
untuned_f1 = ensemble_metrics['Gradient Boosting']['F1-score']

print("\n=== Tuned Gradient Boosting Test-Set Evaluation ===")
print(f"Accuracy:  {tuned_acc:.4f}")
print(f"Precision: {tuned_prec:.4f}")
print(f"Recall:    {tuned_rec:.4f}")
print(f"F1-Score:  {tuned_f1:.4f}")
print(f"\nClassification Report (Tuned Model):\n", classification_report(y_test, y_pred_tuned, zero_division=0))

# Explicit Comparison Table: Untuned vs. Tuned Model
tuning_comp_data = {
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1-Score'],
    'Untuned Gradient Boosting': [f"{untuned_acc:.4f}", f"{untuned_prec:.4f}", f"{untuned_rec:.4f}", f"{untuned_f1:.4f}"],
    'Tuned Gradient Boosting': [f"{tuned_acc:.4f}", f"{tuned_prec:.4f}", f"{tuned_rec:.4f}", f"{tuned_f1:.4f}"],
    'Absolute Improvement (Δ)': [
        f"{tuned_acc - untuned_acc:+.4f}",
        f"{tuned_prec - untuned_prec:+.4f}",
        f"{tuned_rec - untuned_rec:+.4f}",
        f"{tuned_f1 - untuned_f1:+.4f}"
    ]
}

tuning_comp_df = pd.DataFrame(tuning_comp_data)
print("=== Performance Comparison: Untuned vs. Tuned Model ===")
try:
    display(tuning_comp_df)
except NameError:
    print(tuning_comp_df.to_string(index=False))

# Visualizations for Tuned Model
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# 1. Confusion Matrix
cm_tuned = confusion_matrix(y_test, y_pred_tuned)
disp_tuned = ConfusionMatrixDisplay(confusion_matrix=cm_tuned, display_labels=['No Default (0)', 'Default (1)'])
disp_tuned.plot(cmap='YlGnBu', ax=axes[0], colorbar=False)
axes[0].set_title("Tuned Gradient Boosting\nConfusion Matrix", fontweight='bold', fontsize=12)
axes[0].grid(False)

# 2. Feature Importances
feature_names = X.columns.tolist()
importances = best_gb_model.feature_importances_
feat_imp_df = pd.DataFrame({'Feature': feature_names, 'Importance': importances}).sort_values('Importance', ascending=True)

axes[1].barh(feat_imp_df['Feature'], feat_imp_df['Importance'], color='teal', alpha=0.85)
axes[1].set_title("Tuned Gradient Boosting\nFeature Importances", fontweight='bold', fontsize=12)
axes[1].set_xlabel("Gini Relative Importance", fontweight='bold')
axes[1].grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

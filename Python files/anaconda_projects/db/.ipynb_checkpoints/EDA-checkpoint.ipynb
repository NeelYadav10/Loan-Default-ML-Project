### 2.1 Duplicate Removal

```python
# Check and remove duplicate rows
initial_rows = df.shape[0]
duplicate_count = df.duplicated().sum()
print(f"Duplicate rows detected: {duplicate_count}")

df = df.drop_duplicates().reset_index(drop=True)
print(f"Shape before duplicate removal: ({initial_rows}, {df.shape[1]})")
print(f"Shape after duplicate removal:  ({df.shape[0]}, {df.shape[1]})")
```

### 2.2 Missing Value Strategy & Justification
Even though this dataset is clean, robust pipelines require explicit missing value handling:
- **Numerical Features:** Imputed using **Median** via `SimpleImputer(strategy='median')`. Median imputation is robust against skewed numerical distributions and extreme outliers.
- **Categorical Features:** Imputed using **Mode** via `SimpleImputer(strategy='most_frequent')`. Mode imputation preserves valid discrete category levels without introducing artificial non-integer values.

```python
# Step 2.2: Apply SimpleImputer for numerical and categorical features
num_cols = df.select_dtypes(include=['int64', 'float64']).columns.drop('Default', errors='ignore').tolist()
if 'LoanID' in num_cols:
    num_cols.remove('LoanID')
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
if 'LoanID' in cat_cols:
    cat_cols.remove('LoanID')

if len(num_cols) > 0 and df[num_cols].isnull().sum().sum() > 0:
    num_imputer = SimpleImputer(strategy='median')
    df[num_cols] = num_imputer.fit_transform(df[num_cols])

if len(cat_cols) > 0 and df[cat_cols].isnull().sum().sum() > 0:
    cat_imputer = SimpleImputer(strategy='most_frequent')
    df[cat_cols] = cat_imputer.fit_transform(df[cat_cols])

print("Imputation pipeline executed. Total missing values remaining:", df.isnull().sum().sum())
```

### 2.3 Outlier Detection & Handling (IQR Method)
The **Interquartile Range (IQR)** method identifies extreme numerical anomalies beyond statistical thresholds:
$$IQR = Q_3 - Q_1$$
$$\text{Lower Bound} = Q_1 - 1.5 \times IQR, \quad \text{Upper Bound} = Q_3 + 1.5 \times IQR$$

Rather than dropping valuable borrower records, extreme outliers beyond these bounds are **capped** to upper and lower limits to retain total sample size while neutralizing leverage effects.

```python
# Visualize boxplots of key numerical features BEFORE outlier capping
key_num_cols = ['Age', 'Income', 'LoanAmount', 'CreditScore', 'MonthsEmployed', 'InterestRate', 'DTIRatio']
existing_cols = [c for c in key_num_cols if c in df.columns]

plt.figure(figsize=(15, 8))
for i, col in enumerate(existing_cols, 1):
    plt.subplot(2, 4, i)
    sns.boxplot(y=df[col], color='skyblue')
    plt.title(f"Before IQR: {col}", fontsize=11, fontweight='bold')
plt.suptitle("Numerical Feature Distributions (BEFORE Outlier Capping)", y=1.02, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Apply IQR capping on numerical features
df_clean = df.copy()
capping_summary = {}

for col in existing_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers_below = (df_clean[col] < lower_bound).sum()
    outliers_above = (df_clean[col] > upper_bound).sum()
    capping_summary[col] = outliers_below + outliers_above
    
    df_clean[col] = np.clip(df_clean[col], lower_bound, upper_bound)

df = df_clean
print("Capped Outliers Count per Feature:", capping_summary)

plt.figure(figsize=(15, 8))
for i, col in enumerate(existing_cols, 1):
    plt.subplot(2, 4, i)
    sns.boxplot(y=df[col], color='lightgreen')
    plt.title(f"After IQR: {col}", fontsize=11, fontweight='bold')
plt.suptitle("Numerical Feature Distributions (AFTER Outlier Capping)", y=1.02, fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 2.4 Categorical Encoding Strategy
- **LabelEncoder vs. OneHotEncoder:**
  - **LabelEncoder:** Maps categories to ordinal integers ($0, 1, 2, \dots$). Ideal for binary features (`HasMortgage`, `HasDependents`, `HasCoSigner`) or ordinal features (`Education` levels).
  - **OneHotEncoder:** Creates dummy indicator columns for nominal multi-class variables to avoid introducing false numerical ranking.
  - Here, `LabelEncoder` is applied across binary and categorical features to yield a compact matrix representation for modeling.

```python
# Step 2.4: Encode categorical variables using LabelEncoder
cat_columns = ['Education', 'EmploymentType', 'MaritalStatus', 'HasMortgage', 'HasDependents', 'LoanPurpose', 'HasCoSigner']
encoders = {}

for col in cat_columns:
    if col in df.columns:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col].astype(str))
        encoders[col] = le

print("Categorical Encoding complete. Sample encoded values:")
df[cat_columns].head()
```

## Week 3 — Model Creation & Evaluation

### 3.1 Justification: Classification vs. Regression
Loan default prediction is inherently a **classification task** ($y \in \{0, 1\}$) rather than a regression problem for the following reasons:
1. **Discrete Categorical Target:** The target variable `Default` is qualitative and binary (0 = non-default, 1 = default).
2. **Bounded Output Range:** Linear regression models output continuous unbounded predictions ($y \in \mathbb{R}$) which can generate uninterpretable probabilities $<0$ or $>1$.
3. **Loss Function Optimization:** Classification models optimize log-loss (binary cross-entropy) or tree impurity (Gini/Entropy), matching the probability density of decision boundaries.

```python
# Step 3.1: Train Scikit-Learn Logistic Regression Model
t0 = time.time()
sk_model = LogisticRegression(max_iter=1000, random_state=42)
sk_model.fit(X_train_scaled, y_train)
sk_train_time = time.time() - t0

y_pred_sk = sk_model.predict(X_test_scaled)

sk_acc = accuracy_score(y_test, y_pred_sk)
sk_prec = precision_score(y_test, y_pred_sk, zero_division=0)
sk_rec = recall_score(y_test, y_pred_sk, zero_division=0)
sk_f1 = f1_score(y_test, y_pred_sk, zero_division=0)

print("=== Scikit-Learn Logistic Regression Results ===")
print(f"Training Time: {sk_train_time:.4f} seconds")
print(f"Accuracy:  {sk_acc:.4f}")
print(f"Precision: {sk_prec:.4f}")
print(f"Recall:    {sk_rec:.4f}")
print(f"F1-Score:  {sk_f1:.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_sk, zero_division=0))

cm_sk = confusion_matrix(y_test, y_pred_sk)
disp = ConfusionMatrixDisplay(confusion_matrix=cm_sk, display_labels=['No Default (0)', 'Default (1)'])
disp.plot(cmap='Blues')
plt.title("Scikit-Learn Logistic Regression - Confusion Matrix", fontweight='bold')
plt.grid(False)
plt.show()
```

### 3.2 Custom Implementation from Scratch (NumPy Logistic Regression)

The custom model implements **Logistic Regression with Gradient Descent** using pure NumPy:
- **Hypothesis / Sigmoid Activation:**
  $$\hat{y} = \sigma(z) = \frac{1}{1 + e^{-(X w + b)}}$$
- **Binary Cross-Entropy Cost Function:**
  $$J(w, b) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log(\hat{y}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{y}^{(i)}) \right]$$
- **Gradient Descent Updates:**
  $$w \leftarrow w - \alpha \frac{1}{m} X^T (\hat{y} - y), \quad b \leftarrow b - \alpha \frac{1}{m} \sum (\hat{y} - y)$$

```python
# Step 3.2: Complete Custom Logistic Regression Class in Pure NumPy
class LogisticRegressionFromScratch:
    def __init__(self, learning_rate=0.1, num_iterations=1000):
        self.learning_rate = learning_rate
        self.num_iterations = num_iterations
        self.weights = None
        self.bias = None
        self.cost_history = []

    def _sigmoid(self, z):
        z_clipped = np.clip(z, -250, 250)
        return 1.0 / (1.0 + np.exp(-z_clipped))

    def _compute_cost(self, y_true, y_pred):
        m = len(y_true)
        epsilon = 1e-15
        y_pred_clipped = np.clip(y_pred, epsilon, 1.0 - epsilon)
        cost = -(1 / m) * np.sum(y_true * np.log(y_pred_clipped) + (1 - y_true) * np.log(1.0 - y_pred_clipped))
        return cost

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0
        y_arr = y.to_numpy() if hasattr(y, 'to_numpy') else np.array(y)

        for i in range(self.num_iterations):
            linear_model = np.dot(X, self.weights) + self.bias
            y_pred = self._sigmoid(linear_model)

            dw = (1 / n_samples) * np.dot(X.T, (y_pred - y_arr))
            db = (1 / n_samples) * np.sum(y_pred - y_arr)

            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

            if i % 100 == 0:
                c = self._compute_cost(y_arr, y_pred)
                self.cost_history.append(c)

    def predict_proba(self, X):
        linear_model = np.dot(X, self.weights) + self.bias
        return self._sigmoid(linear_model)

    def predict(self, X, threshold=0.5):
        y_predicted = self.predict_proba(X)
        return (y_predicted >= threshold).astype(int)

# Train From-Scratch Model
t0 = time.time()
scratch_model = LogisticRegressionFromScratch(learning_rate=0.1, num_iterations=1000)
scratch_model.fit(X_train_scaled, y_train)
scratch_train_time = time.time() - t0

y_pred_scratch = scratch_model.predict(X_test_scaled)

scratch_acc = accuracy_score(y_test, y_pred_scratch)
scratch_prec = precision_score(y_test, y_pred_scratch, zero_division=0)
scratch_rec = recall_score(y_test, y_pred_scratch, zero_division=0)
scratch_f1 = f1_score(y_test, y_pred_scratch, zero_division=0)

print("=== From-Scratch Logistic Regression Results ===")
print(f"Training Time: {scratch_train_time:.4f} seconds")
print(f"Accuracy:  {scratch_acc:.4f}")
print(f"Precision: {scratch_prec:.4f}")
print(f"Recall:    {scratch_rec:.4f}")
print(f"F1-Score:  {scratch_f1:.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred_scratch, zero_division=0))

cm_scratch = confusion_matrix(y_test, y_pred_scratch)
disp_scratch = ConfusionMatrixDisplay(confusion_matrix=cm_scratch, display_labels=['No Default (0)', 'Default (1)'])
disp_scratch.plot(cmap='Greens')
plt.title("From-Scratch Logistic Regression - Confusion Matrix", fontweight='bold')
plt.grid(False)
plt.show()
```

```python
# Step 3.3: Plot Cost History for From-Scratch Logistic Regression
plt.figure(figsize=(8, 4))
plt.plot(range(0, 1000, 100), scratch_model.cost_history, marker='o', color='darkgreen', linewidth=2)
plt.title("Gradient Descent Loss History (From-Scratch Model)", fontsize=12, fontweight='bold')
plt.xlabel("Iteration")
plt.ylabel("Binary Cross-Entropy Loss")
plt.grid(True)
plt.show()
```

### 3.3 Performance Comparison & Discussion

1. **Accuracy & Metric Parity:**
   - Both models achieve virtually identical accuracy (**~88.51%** vs **~88.50%**) on the test dataset.
   - Their precision, recall, and F1-scores align closely, proving that the mathematical formulation of our custom NumPy gradient descent implementation matches Scikit-Learn's underlying optimization logic.

2. **Execution Speed & Computational Efficiency:**
   - **Scikit-Learn** trained in **~0.16 seconds**.
   - **From-Scratch NumPy** trained in **~7.5 seconds** (nearly **45x slower**).
   - *Reason:* Scikit-Learn wraps highly optimized C/Fortran routines (`LIBLINEAR` / `L-BFGS`) and utilizes quasi-Newton second-order Hessian approximations, whereas pure Python/NumPy gradient descent performs un-parallelized first-order updates step-by-step.

3. **Why Machine Learning Libraries are Used in Practice:**
   - **Numerical Stability:** Production libraries prevent gradient explosion and floating-point underflow/overflow using techniques like log-sum-exp stabilization and dynamic learning rate line searches.
   - **Advanced Optimization Solvers:** Solvers like L-BFGS, SAG, and SAGA converge in fewer iterations than standard first-order gradient descent.
   - **Built-in Regularization & Hyperparameter Tuning:** Automatic $L_1$ and $L_2$ penalty handling prevents overfitting without custom gradient algebra.
   - **Scalability & Production Infrastructure:** Integration with multi-threading, C++ backends, pipelines, and deployment frameworks makes libraries indispensable for industry applications.
